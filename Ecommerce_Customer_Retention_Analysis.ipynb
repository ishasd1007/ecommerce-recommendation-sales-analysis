```python
# Remove records without a usable customer identity
sales_rfm = sales.dropna(
    subset=["customer_unique_id"]
).copy()

reference_date = (
    sales_rfm["order_purchase_timestamp"].max()
    + pd.Timedelta(days=1)
)

rfm = sales_rfm.groupby("customer_unique_id").agg(
    Last_Purchase=("order_purchase_timestamp", "max"),
    Frequency=("order_id", "nunique"),
    Monetary=("price", "sum")
)

rfm["Recency"] = (
    reference_date - rfm["Last_Purchase"]
).dt.days

rfm = rfm[
    ["Recency", "Frequency", "Monetary"]
]

rfm.head()
```

```python
# Remove records without a usable customer identity
sales_rfm = sales.dropna(
    subset=["customer_unique_id"]
).copy()

# Reference date
reference_date = (
    sales_rfm["order_purchase_timestamp"].max()
    + pd.Timedelta(days=1)
)

# Calculate RFM for ALL customers
rfm = sales_rfm.groupby("customer_unique_id").agg(
    Last_Purchase=("order_purchase_timestamp", "max"),
    Frequency=("order_id", "nunique"),
    Monetary=("price", "sum")
)

# Calculate Recency
rfm["Recency"] = (
    reference_date - rfm["Last_Purchase"]
).dt.days

# Keep required columns
rfm = rfm[["Recency", "Frequency", "Monetary"]]

# Display ALL customers
print("Total customers:", len(rfm))
display(rfm)
```

```python
rfm.to_csv("customer_rfm_all.csv", index=True)
```

```python
# Assign scores from 1 to 5
# Recency: lower days = better score
rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

# Frequency: higher orders = better score
rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Monetary: higher spending = better score
rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Combined RFM score
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str)
    + rfm["F_Score"].astype(str)
    + rfm["M_Score"].astype(str)
)

display(rfm)
```

```python
def segment_customer(row):
    if row["R_Score"] >= 4 and row["F_Score"] >= 4:
        return "Loyal / Best Customers"
    elif row["R_Score"] >= 4 and row["F_Score"] <= 2:
        return "Recent Customers"
    elif row["R_Score"] <= 2 and row["F_Score"] >= 3:
        return "At Risk Customers"
    elif row["R_Score"] <= 2 and row["F_Score"] <= 2:
        return "Lost / Inactive Customers"
    else:
        return "Regular Customers"

rfm["Customer_Segment"] = rfm.apply(
    segment_customer, axis=1
)

display(rfm)
```

```python
segment_summary = (
    rfm.groupby("Customer_Segment")
    .agg(
        Total_Customers=("Customer_Segment", "size"),
        Avg_Recency=("Recency", "mean"),
        Avg_Frequency=("Frequency", "mean"),
        Avg_Monetary=("Monetary", "mean")
    )
    .round(2)
    .sort_values("Total_Customers", ascending=False)
)

display(segment_summary)
```

```python
# Convert order purchase timestamp to datetime
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"],
    errors="coerce"
)

# Keep delivered orders for consistent retention analysis
delivered_orders = orders[
    orders["order_status"] == "delivered"
].copy()

# Merge customer identity with orders
customer_orders = delivered_orders.merge(
    customers[["customer_id", "customer_unique_id"]],
    on="customer_id",
    how="left"
)

# Remove records without customer identity
customer_orders = customer_orders.dropna(
    subset=["customer_unique_id", "order_purchase_timestamp"]
)

print("Delivered orders:", len(customer_orders))
print(
    "Unique customers:",
    customer_orders["customer_unique_id"].nunique()
)

display(customer_orders.head())
```

```python
# All delivered orders with customer identity
customer_orders = delivered_orders.merge(
    customers[["customer_id", "customer_unique_id"]],
    on="customer_id",
    how="left"
)

customer_orders = customer_orders.dropna(
    subset=["customer_unique_id", "order_purchase_timestamp"]
).copy()

customer_orders["order_purchase_timestamp"] = pd.to_datetime(
    customer_orders["order_purchase_timestamp"],
    errors="coerce"
)

customer_orders = customer_orders.dropna(
    subset=["order_purchase_timestamp"]
)

# Purchase month
customer_orders["Order_Month"] = (
    customer_orders["order_purchase_timestamp"].dt.to_period("M")
)

# First purchase month for every customer
customer_orders["Cohort_Month"] = (
    customer_orders.groupby("customer_unique_id")["Order_Month"]
    .transform("min")
)

# Months since first purchase
customer_orders["Cohort_Index"] = (
    (customer_orders["Order_Month"].dt.year -
     customer_orders["Cohort_Month"].dt.year) * 12
    + customer_orders["Order_Month"].dt.month
    - customer_orders["Cohort_Month"].dt.month
    + 1
)

print("Total eligible customers:",
      customer_orders["customer_unique_id"].nunique())
```

```python
cohort_data = (
    customer_orders.groupby(
        ["Cohort_Month", "Cohort_Index"]
    )["customer_unique_id"]
    .nunique()
    .reset_index(name="Customers")
)

cohort_counts = cohort_data.pivot(
    index="Cohort_Month",
    columns="Cohort_Index",
    values="Customers"
)

# All cohort months and all available customer counts
cohort_sizes = cohort_counts[1]

retention = cohort_counts.div(cohort_sizes, axis=0) * 100

print("Total cohort months:", len(retention))
print("Retention table shape:", retention.shape)

display(retention.round(2))  # Full table, not just 5 rows
```

```python
# Count unique customers for each cohort month
# and month since first purchase

cohort_data = (
    customer_orders.groupby(
        ["Cohort_Month", "Cohort_Index"]
    )["customer_unique_id"]
    .nunique()
    .reset_index(name="Customers")
)

# Create cohort table
cohort_counts = cohort_data.pivot(
    index="Cohort_Month",
    columns="Cohort_Index",
    values="Customers"
)

# Calculate retention percentage
cohort_sizes = cohort_counts[1]

retention = cohort_counts.div(
    cohort_sizes, axis=0
) * 100

print("Total cohort months:", len(retention))
print("Cohort table dimensions:", retention.shape)

display(retention.round(2))
```

```python
category_translation = pd.read_csv(
    DATA_DIR / "product_category_name_translation.csv"
)

display(category_translation.head())
print("Total rows:", len(category_translation))
```

```python
# Merge order items with product details
product_analysis = items.merge(
    products,
    on="product_id",
    how="left"
)

# Add English product category names
product_analysis = product_analysis.merge(
    category_translation,
    on="product_category_name",
    how="left"
)

# Use English category where available
product_analysis["category"] = (
    product_analysis["product_category_name_english"]
    .fillna(product_analysis["product_category_name"])
    .fillna("Unknown")
)

# Calculate category-level sales and orders
category_summary = (
    product_analysis.groupby("category")
    .agg(
        Total_Items=("order_item_id", "count"),
        Unique_Orders=("order_id", "nunique"),
        Total_Sales=("price", "sum"),
        Total_Freight=("freight_value", "sum")
    )
    .sort_values("Total_Sales", ascending=False)
    .round(2)
)

# Display all categories
print("Total categories:", len(category_summary))
display(category_summary)
```

```python
# Keep delivered orders for customer purchase analysis
delivered_orders = orders[
    orders["order_status"] == "delivered"
].copy()

# Attach unique customer identity to each order
customer_analysis = delivered_orders.merge(
    customers[["customer_id", "customer_unique_id"]],
    on="customer_id",
    how="left"
)

# Remove records without customer identity
# Include all orders, not only delivered orders
customer_analysis = orders.merge(
    customers[["customer_id", "customer_unique_id"]],
    on="customer_id",
    how="left"
)

# Remove records without usable customer identity
customer_analysis = customer_analysis.dropna(
    subset=["customer_unique_id"]
).copy()

customer_analysis["order_purchase_timestamp"] = pd.to_datetime(
    customer_analysis["order_purchase_timestamp"],
    errors="coerce"
)

# Summarize all available orders for each unique customer
customer_summary = (
    customer_analysis.groupby("customer_unique_id")
    .agg(
        Total_Orders=("order_id", "nunique"),
        First_Order=("order_purchase_timestamp", "min"),
        Last_Order=("order_purchase_timestamp", "max"),
        Order_Statuses=("order_status", "nunique")
    )
    .reset_index()
)

customer_summary["Customer_Type"] = np.where(
    customer_summary["Total_Orders"] > 1,
    "Repeat Customer",
    "One-Time Customer"
)

print("Total unique customers:", len(customer_summary))
display(customer_summary)  # Display all customers
```
